```python
import requests

# Proxy server details
PROXY_SERVER_URL = "http://192.168.56.1:5050"  # Replace with the actual proxy server URL
API_KEY = "your_api_key_here"  # Replace with the API key if required

# API endpoints
CONNECT_ENDPOINT = "/connect"
CAPTURE_ENDPOINT = "/capture"
DISCONNECT_ENDPOINT = "/disconnect"

def connect_to_scanner():
    """Connect to the Mantra device via the proxy server."""
    url = f"{PROXY_SERVER_URL}{CONNECT_ENDPOINT}"
    headers = {
        "Authorization": f"Bearer {API_KEY}",  # If authentication is required
        "Content-Type": "application/json",
    }
    response = requests.post(url, headers=headers)
    if response.status_code == 200:
        print("Connected to the scanner successfully!")
        return True
    else:
        print(f"Failed to connect to the scanner. Error: {response.text}")
        return False

def capture_fingerprint():
    """Capture fingerprint data from the Mantra device."""
    url = f"{PROXY_SERVER_URL}{CAPTURE_ENDPOINT}"
    headers = {
        "Authorization": f"Bearer {API_KEY}",  # If authentication is required
        "Content-Type": "application/json",
    }
    response = requests.post(url, headers=headers)
    if response.status_code == 200:
        fingerprint_data = response.json().get("data")
        print("Fingerprint captured successfully!")
        print("Fingerprint Data:", fingerprint_data)
        return fingerprint_data
    else:
        print(f"Failed to capture fingerprint. Error: {response.text}")
        return None

def disconnect_scanner():
    """Disconnect from the Mantra device."""
    url = f"{PROXY_SERVER_URL}{DISCONNECT_ENDPOINT}"
    headers = {
        "Authorization": f"Bearer {API_KEY}",  # If authentication is required
        "Content-Type": "application/json",
    }
    response = requests.post(url, headers=headers)
    if response.status_code == 200:
        print("Scanner disconnected successfully!")
    else:
        print(f"Failed to disconnect the scanner. Error: {response.text}")

def test_scanner():
    """Test the Mantra scanner via the proxy server."""
    if connect_to_scanner():
        fingerprint_data = capture_fingerprint()
        if fingerprint_data:
            # Process the fingerprint data (e.g., save to database or verify)
            print("Processing fingerprint data...")
        disconnect_scanner()

if __name__ == "__main__":
    test_scanner()
```
